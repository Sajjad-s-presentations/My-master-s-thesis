```python
def translate_to_persian(text):
    if pd.isna(text) or text == '':
        return text
    try:
        translator = GoogleTranslator(source='auto', target='fa')
        return translator.translate(text)
    except Exception as e:
        print(f"خطا در ترجمه: {e}")
        return text
```
